# Population Dynamics Module Test

## Testing Population Modeling with Immigration/Emigration Flows

This notebook validates the population dynamics module with:
- 🏗️ **Multidimensional Population Stocks**: Age, gender, and income dimensions
- 🌊 **Immigration/Emigration Flows**: Demographic-specific migration patterns
- 👶 **Birth/Death Flows**: Natural population dynamics
- 🧮 **Calculator Components**: Net migration and population change calculations
- 📊 **Basic Visualizations**: Population trends and demographic analysis

```python
# Import required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import yaml
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

# Set plotting style
plt.style.use('seaborn-v0_8')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 11

print("✅ Libraries imported successfully!")
```

```python
# Add the sd_toolkit to the path
import sys
import os
sys.path.insert(0, os.path.join('..', '..', '..', 'sd_toolkit'))

# Import System Dynamics Toolkit components
from sd_toolkit.config import YAMLSystemBuilder, TemplateManager, rate_registry
from sd_toolkit.engine.system import SystemModel
from sd_toolkit.engine.elements import Stock, Flow, Auxiliary, Calculator
from sd_toolkit.data.loader import SpatioTemporalData
from sd_toolkit.analysis.plotting import SystemPlotter
from sd_toolkit.core.units import Q_, unit_registry

print("✅ System Dynamics Toolkit loaded successfully!")
print("🏗️ Ready for population dynamics module testing")
print(f"📊 Available rate functions: {len(rate_registry.list_functions())}")
```

```python
# Load model structure and scenario parameters
print("📂 Loading population dynamics configuration...")

# Load model structure
model_structure_path = Path('model_structure.yaml')
with open(model_structure_path, 'r') as file:
    model_structure = yaml.safe_load(file)

print(f"✅ Model structure loaded: {model_structure['model']['name']}")

# Load scenario parameters
scenario_path = Path('scenario_parameters.yaml')
with open(scenario_path, 'r') as file:
    scenario_params = yaml.safe_load(file)

print(f"✅ Scenario parameters loaded: {scenario_params['scenario']['name']}")

# Display model dimensions
print("\n🏗️ Model Dimensions:")
for dim_name, dim_info in model_structure['dimensions'].items():
    print(f"  • {dim_name}: {dim_info['labels']} (size: {dim_info['size']})")
```

```python
# Merge model structure with scenario parameters
print("🔧 Building population dynamics model...")

# Merge configurations
full_config = model_structure.copy()
full_config['constants'] = scenario_params['constants']

# Create YAML system builder
builder = YAMLSystemBuilder()

# Build the model
try:
    population_model = builder.build_from_dict(full_config)
    print(f"✅ Model built successfully: {population_model.name}")
    print(f"📊 Model elements: {len(population_model.elements)}")
    print(f"📦 Stocks: {len(population_model.stocks)}")
    print(f"🌊 Flows: {len(population_model.flows)}")
    print(f"🧮 Calculators: {len(population_model.calculators)}")
    print(f"🔧 Auxiliaries: {len(population_model.auxiliaries)}")
    
except Exception as e:
    print(f"❌ Error building model: {e}")
    population_model = None
```

```python
# Validate model structure and run basic simulation
if population_model is not None:
    print("🚀 Running population dynamics simulation...")
    
    try:
        # Run simulation
        results = population_model.simulate(time_horizon=20, dt=0.25)
        
        print(f"✅ Simulation completed successfully!")
        print(f"📊 Results shape: {results.shape if hasattr(results, 'shape') else 'N/A'}")
        print(f"⏱️ Time steps: {len(results) if hasattr(results, '__len__') else 'N/A'}")
        
        # Display basic statistics
        if hasattr(results, 'columns'):
            print(f"📈 Variables tracked: {list(results.columns)}")
        
    except Exception as e:
        print(f"❌ Simulation error: {e}")
        results = None
else:
    print("❌ Cannot run simulation - model not built")
    results = None
```
